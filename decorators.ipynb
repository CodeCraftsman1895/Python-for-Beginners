## Decorators in Python

Decorators in Python are used to add or modify the behavior of a function (or method)
without changing its original source code. They work by wrapping one function inside 
another function,allowing additional functionality to be executedbefore or after the 
original function runs.

```python
def white():
    print("white")

def colors(func):
    def wrapper():
        print("saffron")
        func()
        print("green")
    return wrapper
    

flag = colors(white)
flag()
```

```python
def colors(func):
    def wrapper():
        print("saffron")
        func()
        print("green")
    return wrapper

@colors
def white():
    print("white")

white()
```

```python
def operation(a,b,c,d):
    result = (a/b)*(c/d)
    return result

def dec_func(f):
    def wrapper(a,b,c,d):
        if a<b:
            a,b = b,a
        if c<d:
            c,d = d,c
        return f(a,b,c,d)
    return wrapper

oper = dec_func(operation)

result = oper(3,6,4,8)
print(result)
```
