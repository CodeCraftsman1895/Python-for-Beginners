```python
str1 = input("Enter a string: ")
str2 = ""
for i in range(len(str1)-1,-1,-1):
    str2 += str1[i]

if str1 == str2:
    print("palindrome")
else:
    print("Not a palindrome")
```

```python
string = input("Enter a sentence: ")
words = string.split()
lw = ""
sw = " "*len(string)
for word in words:
    if len(word) > len(lw):
        lw = word
    if len(word) < len(sw):
        sw = word


print(sw)
print(lw)
```
