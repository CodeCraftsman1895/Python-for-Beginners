```python
# 1) First occurence and last occurrence of given number in list (index print cheyali
# 2) Write a program to print x index value of a list
def presence(ele, lst):
    index = []
    for i in range(len(lst)):
        if lst[i] == ele:
            index.append(i)
    return index, index[0], index[-1]

lst = list(map(int, input("Enter the elements in a single string").split()))
ele = int(input("Enter the element you wish to find it's place: "))
index, first, last = presence(ele, lst)
print(f'the first occurence is at index {first} and last occurence is at index {last}')
print(f"All the indices where the element occurs{index}")            
    
```

```python
# 
lst = list(map(int, input().split()))
unique = []
for i in lst:
    if i not in unique:
        unique.append(i)

count = [0]*len(unique)
for i in range(len(lst)):
    if lst[i] in unique:
        count[unique.index(lst[i])] += 1
non_dup = []
for i in range(len(count)):
    if count[i]>1:
        print(f'The element {unique[i]} is repeated {count[i]} times')
    if count[i] == 1:
        non_dup.append(unique[i])

print(f'The unique elements that are not repeated are {non_dup}')


        
```
