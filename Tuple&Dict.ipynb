```python
# dictionary
dct = {
    'Key': "value",
    "Name": "Dinesh"
    
} # this is how we create a dictionary and it has keys and corresponding values
```

```python
dct.get("value")
```

```python
dct.keys() # prints the keys of the dct
```

```python
dct.values() # prints the values of the dct
```

```python
dct.items() # prints both key and value pairs as a list of tuples
```

```python
dct.pop("Key","Not Found") # ("Not Found") is the default for the pop method and it is returned if the specified key is not found
```
