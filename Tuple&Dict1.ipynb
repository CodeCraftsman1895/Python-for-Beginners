# Dictionary

```python
abhi1 = {
    "Mango": 6,
    "banana": 2,
    "Sapota": 9,
    "Apple" : 5,
    "grapes" : 15
} # dictionary is a data structure written in curly braces with key and value pairs
```

```python
abhi1
```

```python
abhi1["Mango"] # way to access elements using keys
```

```python
abhi1["banana"]
```

```python
abhi1.keys() # gives all the keys present in a dictionary
```

```python
abhi1.values() # gives all the values present in a dictionary
```

```python
abhi1.items() # gives the key value pairs of a dictionary
```

```python
for i,j in abhi1.items():
    print(i,j) # one of the way to iterate through a dictionary
```

```python
abhi1["orange"] = 8 # adding new elements
```

```python
abhi1
```

```python
len(abhi1) # to find the length of the dictionary
```

```python
abhi1[1] = 2 # data type is not specific for dictionary we can give any data type for keys as well as values
```

```python
abhi1
```

```python
del abhi1[1] # remove the key value pair
```

```python
abhi1
```

```python
fruits = {
    "Name" : ["Mango", "Banana", "Sapota", "Orange", "kiwi", "Lichi"],
    "Quantity": [5, 6, 3, 7, 4,9]
} # we have assigned multiple values for a single key using list
```

```python
# you will learn in future
import pandas as pd
df = pd.DataFrame(fruits)
df
```
