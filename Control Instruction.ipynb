```python
# flag based loops
J = True
while J:
    phone = input("Please enter your name")
    if phone == "Dinesh":
        print("Anonymous")
        J = False
```

```python
lst = []
Flag = False
for i in range(100):
    if i >50:
        Flag = True
    if i%3 == 0 and Flag == False:
        lst.append(i)
    if i%5 == 0 and Flag == True:
        lst.append(i)

lst
```

```python
n = 5
fact = 1
if n == 0:
    fact = 1
elif n < 0:
    print("Give a valid number")
    fact = 0
else:
    for i in range(n, 0, -1):
        fact = fact * (i)
fact
```
