# Classes and Objects

````python
# defining a class
class Animal():
    name = "Monkey D. Luffy"
#creating an object
a1 = Animal()
print(a1.name)
````

````python
# constructor function
# a default function of each class which executes when getting initiated even we don't explicitly define it
class Animal():
    def __init__(self, name):
        self.name = name

a1 = Animal("Monkey D. Luffy")
print(a1.name)
````

````python
# self parameter is reference to the current instance of the class (object)
# method is a function defined inside a class
class Animal():
    def __init__(self, name):
        self.name = name

    def Intro(self):
        print("I’m " + self.name + "! The man who’s gonna be King of the Pirates!")

a1 = Animal("Monkey D. Luffy")
print(a1.Intro())
````

````python
# simple implementation to show an example of abstraction
class TV():
    def power_on(self):
        print("TV is ON")
r = TV()
r.power_on()
````

This sure tells how abstraction works but the class is not abstract class and the method is not abstract method

````python
# Abstraction Example
from abc import ABC, abstractmethod
class Vehicle(ABC): # this marks the class as an Abstract Base Class (ABC)
    pass

v = Vehicle() # this is valid since it does not have atleast one single abstract method

````

````python
from abc import ABC, abstractmethod
class Vehicle(ABC): 
    @abstractmethod  # this marks which methods the children of class must implement
    def start(self):
        pass
try:
    v = Vehicle() # not possible because it has a method not implemented and v.start() does nothing
except TypeError as e:
    print(e)
````

````python
from abc import ABC, abstractmethod
class Vehicle(ABC): 
    @abstractmethod 
    def start(self):
        pass

class Car(Vehicle):
    def start(self):
        print("The car has started")

c1 = Car()
c1.start()
````

## Encapsulation - wrapping data and functions into single unit and hiding of data
### In python Encapsulation is achieved by access modifiers 
- public (no underscore before the attribute)
    ```python
        class Student:
            def __init__(self, name):
                self.name = name  # Public attribute
    
        s = Student("Dinesh")
        print(s.name)  # Accessible    
    ```

    
- private (Double underscore)
    ```python
        class Student:
            def __init__(self, name):
                self.__name = name  # Private attribute
        
            def get_name(self):
                return self.__name
        
        s = Student("Dinesh")
        
        # print(s.__name)  # Raises AttributeError
        print(s.get_name())  # Correct way   
    ```

    
- protected (Single underscore)
    ```python
        class Student:
            def __init__(self, name):
                self._name = name  # Protected attribute
    
        s = Student("Dinesh")
        print(s.name)  # Possible, but discouraged   
    ```

we can use setter and getter methods to access private attributes

## Inheritance - one class inherits the features of another class
- Single Inheritance
- Multi-level Inheritance
- Multiple-Inheritance
- Heirarchial Inheritance

### 1.Single Inheritance

````python
class A():
    def __init__(self):
        print("This is class A")
    def fn1(self):
        print("This is a fn in class A")


class B(A):
    def __init__(self):
        print("Class B inherits class A")


b = B()
b.fn1()
# A single parent and a child class
````

### 2.Multi-level Inheritance

````python
class A():
    def __init__(self):
        print("This is class A")
    def fn1(self):
        print("This is a fn in class A")


class B(A):
    def __init__(self):
        print("Class B inherits class A")
    def fn2(self):
        print("This is a fn in class B")

class C(B):
     def __init__(self):
        print("Class C inherits class B")

c = C()
c.fn1()
c.fn2()
````

### 3.Multiple-Inheritance

````python
class A():
    def __init__(self):
        print("This is class A")
    def fn1(self):
        print("This is a fn in class A")


class B():
    def __init__(self):
        print("This is class B")
    def fn2(self):
        print("This is a fn in class B")

class C(A,B):
     def __init__(self):
        print("Class C inherits classes A and B")

c = C()
c.fn1()
c.fn2()
````

### 4.Heirarchial Inheritance

````python
class A():
    def __init__(self):
        print("This is class A")
    def fn1(self):
        print("This is a fn in class A")


class B(A):
    def __init__(self):
        print("Class B inherits class A")
    def fn2(self):
        print("This is a fn in class B")

class C(A):
    def __init__(self):
        print("Class C inherits class A")

c = C()
c.fn1()
try:
    c.fn2()
except AttributeError as e:
    print(e)
````

### super() method
- it is used to access methods of parent class

````python
class Bike():
    def __init__ (self):
        print("The Brand is Royal Enfield")
class RoyalEnfield(Bike):
    def __init__(self, model):
        self.model = model
        print(model)
        super().__init__()
bike = RoyalEnfield("Himalayan") #how to not print this brand but print the brand in the parent class
````

### Method Resolution Order (MRO)

Method Resolution Order (MRO) is the order in which Python searches for methods and attributes when they are called on an object.

````python
class A():
    def __init__(self):
        print("This is class A")
    def fn1(self):
        print("This is a fn in class A")


class B():
    def __init__(self):
        print("This is class B")
    def fn2(self):
        print("This is a fn in class B")

class C(A,B):
     def __init__(self):
        print("Class C inherits classes A and B")

c = C()
c.fn1()
c.fn2()
C.__mro__
````

mro checks the child class first if the method is not present in child class it searches according to mro order

## Polymorphism - an object can have multiple forms

eg: 
- operator overloading

1
- 10 + 20 = 30
- "hell" + "o" = "hello"

here + operator can be used as an addition operator as well as concatenation operator

2
- x = 30
- x = "Age"
x can be used to store both int and string values

````python
# Method Overriding
class BigMom():
    def laugh(self):
        print("Mamamama")
class BlackBeard(BigMom):
    def laugh(self):
        print("Zehaahahaha")
class Brook(BlackBeard):
    def laugh(self):
        print("Yohohohoho")
music = Brook()
music.laugh()

# Though all the classes have the same laugh(self) method with same parameters but the child class brook dominates them all 
````

method overloading is having same method name with different parameters and it is not supported in python so this can solved using variable arguments or default arguments
